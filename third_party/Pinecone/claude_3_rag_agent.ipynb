# LangChain v1을 사용한 Claude 3 RAG Agent

LangChain v1은 많은 변화를 가져왔으며, LangChain 버전 `0.0.3xx`와 `0.1.x`를 비교하면 권장되는 작업 방식에 많은 차이가 있습니다. Agent의 경우 특히 그렇습니다.

Agent를 초기화하고 사용하는 방식은 일반적으로 과거보다 명확해졌습니다. 여전히 많은 abstraction이 있지만 agent logic 자체에 더 가까이 접근할 수 있고 그렇게 하도록 권장됩니다. 처음에는 다소 혼란스러울 수 있지만 이해하고 나면 새 logic이 이전 버전보다 훨씬 명확할 수 있습니다.

이 예제에서는 LangChain v1으로 RAG agent를 구축하겠습니다. LLM에는 Claude 3, 지식 embedding에는 Voyage AI, 지식 retrieval에는 Pinecone을 사용합니다.

RAG를 사용하는 agent에 가장 먼저 필요한 것은 지식을 가져올 곳입니다. Hugging Face Datasets의 [`jamescalam/ai-arxiv2-chunks`](https://huggingface.co/datasets/jamescalam/ai-arxiv2-chunks)에서 제공하는 AI ArXiv dataset v2를 사용하겠습니다.

_참고: 미리 chunk로 나뉜 dataset을 사용합니다. 원본 버전은 [`jamescalam/ai-arxiv2`](https://huggingface.co/datasets/jamescalam/ai-arxiv2)를 참조하세요._

````python
from datasets import load_dataset

dataset = load_dataset("jamescalam/ai-arxiv2-chunks", split="train[:20000]")
dataset
````

## Knowledge Base 구축

Knowledge base를 구축하려면 _두 가지_가 필요합니다:

1. Embedding. Voyage AI의 embedding 모델을 사용하는 `VoyageEmbeddings`를 사용하며, [API key](https://dash.voyageai.com/api-keys)가 필요합니다.
2. Embedding을 저장하고 query할 vector database. 역시 [무료 API key](https://app.pinecone.io)가 필요한 Pinecone을 사용합니다.

먼저 Voyage AI 연결을 초기화하고 embedding용 `embed` object를 정의합니다:

````python
from langchain_community.embeddings import VoyageEmbeddings

embed = VoyageEmbeddings(voyage_api_key=VOYAGE_API_KEY, model="voyage-2")
````

그런 다음 Pinecone 연결을 초기화합니다:

````python
from pinecone import Pinecone

# Client 설정
pc = Pinecone(api_key=PINECONE_API_KEY)
````

이제 index를 배포할 cloud provider와 region을 정의할 수 있는 index specification을 설정합니다. [사용 가능한 모든 provider와 region 목록은 여기](https://docs.pinecone.io/docs/projects)에서 확인할 수 있습니다.

````python
from pinecone import ServerlessSpec

spec = ServerlessSpec(cloud="aws", region="us-west-2")
````

Index를 생성하기 전에 Voyage AI embedding 모델의 dimension이 필요합니다. Embedding을 생성하고 길이를 확인하면 쉽게 알 수 있습니다:

````python
vec = embed.embed_documents(["ello"])
len(vec[0])
````

이제 embedding dimension과 모델에 호환되는 metric(cosine 또는 dotproduct)을 사용하여 index를 생성합니다. Index 초기화 시 spec도 전달합니다.

````python
import time

index_name = "claude-3-rag"

# Index가 이미 있는지 확인(처음이라면 없어야 함)
if index_name not in pc.list_indexes().names():
    # 없으면 index 생성
    pc.create_index(
        index_name,
        dimension=len(vec[0]),  # Voyage 모델의 dimension
        metric="dotproduct",
        spec=spec,
    )
    # Index 초기화 대기
    while not pc.describe_index(index_name).status["ready"]:
        time.sleep(1)

# Index에 연결
index = pc.Index(index_name)
time.sleep(1)
# Index 통계 보기
index.describe_index_stats()
````

### Index에 데이터 채우기

이제 knowledge base에 데이터를 채울 준비가 되었습니다. `embed` 보조 함수를 사용하여 문서를 embedding한 뒤 index에 추가하겠습니다.

각 record의 metadata도 포함하겠습니다.

````python
from tqdm.auto import tqdm

# Dataset을 pandas dataframe으로 다루는 편이 더 쉬움
data = dataset.to_pandas()

batch_size = 100

for i in tqdm(range(0, len(data), batch_size)):
    i_end = min(len(data), i + batch_size)
    # 데이터 batch 가져오기
    batch = data.iloc[i:i_end]
    # 각 chunk의 unique id 생성
    ids = [f"{x['doi']}-{x['chunk-id']}" for i, x in batch.iterrows()]
    # Embedding할 텍스트 가져오기
    texts = [x["chunk"] for _, x in batch.iterrows()]
    # 텍스트 embedding
    embeds = embed.embed_documents(texts)
    # Pinecone에 저장할 metadata 가져오기
    metadata = [
        {"text": x["chunk"], "source": x["source"], "title": x["title"]}
        for i, x in batch.iterrows()
    ]
    # Pinecone에 추가
    index.upsert(vectors=zip(ids, embeds, metadata, strict=False))
````

Agent가 ArXiv 논문을 검색할 때 사용할 tool을 생성합니다:

````python
from langchain.agents import tool


@tool
def arxiv_search(query: str) -> str:
    """AI, machine learning, data science 또는 arXiv 논문으로 답할 수 있는
    다른 기술 질문에 답할 때 이 tool을
    사용하세요.
    """
    # Query vector 생성
    xq = embed.embed_query(query)
    # 검색 수행
    out = index.query(vector=xq, top_k=5, include_metadata=True)
    # 결과를 문자열로 정리
    results_str = "\n\n".join([x["metadata"]["text"] for x in out["matches"]])
    return results_str


tools = [arxiv_search]
````

Agent가 이 tool을 사용할 때 다음과 같이 실행합니다:

````python
print(arxiv_search.run(tool_input={"query": "can you tell me about llama 2?"}))
````

## XML Agent 정의

XML agent는 주로 Anthropic 모델을 지원하도록 구축되었습니다. Anthropic 모델은 `<input>{some input}</input`과 같은 XML tag를 사용하도록 학습되었으며, tool을 사용할 때는 다음을 사용합니다:

```
<tool>{tool name}</tool>
<tool_input>{tool input}</tool_input>
```

이는 일반적인 ReAct agent가 생성하는 형식과 매우 다르며, 그 형식은 Anthropic 모델에서 지원이 상대적으로 부족합니다.

XML agent를 생성하려면 `prompt`, `llm`, `tools` 목록이 필요합니다. LangChain hub에서 대화형 XML agent용으로 미리 만들어진 prompt를 다운로드할 수 있습니다.

````python
from langchain import hub

prompt = hub.pull("hwchase17/xml-agent-convo")
prompt
````

LLM에 tool 사용 방법을 설명할 때 prompt 전반에 XML 형식이 사용되는 것을 볼 수 있습니다.

다음으로 Anthropic 연결을 초기화합니다. 이를 위해 [Claude API key](https://console.anthropic.com/)가 필요합니다.

````python
from langchain_anthropic import ChatAnthropic

# chat completion llm
llm = ChatAnthropic(
    ANTHROPIC_API_KEY=ANTHROPIC_API_KEY,
    model_name="claude-opus-4-8",  # 속도를 위해 "opus" -> "sonnet"으로 변경
    temperature=0.0,
)
````

Agent가 실행될 때 사용자 input 텍스트인 `input` 하나를 제공합니다. 그러나 agent logic 내부에서는 tool 정보를 포함한 *agent_scratchpad* object도 전달됩니다. 이 정보를 LLM에 전달하려면 위에서 설명한 XML 형식으로 변환해야 하므로 이를 처리할 `convert_intermediate_steps` 함수를 정의합니다.

````python
def convert_intermediate_steps(intermediate_steps):
    log = ""
    for action, observation in intermediate_steps:
        log += (
            f"<tool>{action.tool}</tool><tool_input>{action.tool_input}"
            f"</tool_input><observation>{observation}</observation>"
        )
    return log
````

Tool을 `tool_name: tool_description`이 포함된 문자열로 parsing해야 합니다. 이는 `convert_tools` 함수로 처리합니다.

````python
def convert_tools(tools):
    return "\n".join([f"{tool.name}: {tool.description}" for tool in tools])
````

모든 준비가 끝났으므로 [**L**ang**C**hain **E**xpression **L**anguage (LCEL)](https://www.pinecone.io/learn/series/langchain/langchain-expression-language/)로 agent object를 초기화할 수 있습니다. `llm.bind(stop=[...])`로 LLM이 생성을 _중지_해야 할 때를 지정하고, 마지막으로 `XMLAgentOutputParser` object로 agent output을 parsing합니다.

````python
from langchain.agents.output_parsers import XMLAgentOutputParser

agent = (
    {
        "input": lambda x: x["input"],
        # "chat_history" 없이는 tool 사용 시 이전 상호 작용의 context가 없음
        "chat_history": lambda x: x["chat_history"],
        "agent_scratchpad": lambda x: convert_intermediate_steps(x["intermediate_steps"]),
    }
    | prompt.partial(tools=convert_tools(tools))
    | llm.bind(stop=["</tool_input>", "</final_answer>"])
    | XMLAgentOutputParser()
)
````

`agent` object를 초기화한 뒤 원래의 `tools` 목록과 함께 `AgentExecutor` object에 전달합니다:

````python
from langchain.agents import AgentExecutor

agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)
````

이제 `invoke` method를 통해 agent를 사용할 수 있습니다:

````python
user_msg = "can you tell me about llama 2?"

out = agent_executor.invoke({"input": user_msg, "chat_history": ""})

print(out["output"])
````

꽤 좋아 보이지만 현재 agent는 _stateless_이므로 대화하기 어렵습니다. 다양한 방법으로 memory를 제공할 수 있지만 가장 쉬운 방법 중 하나는 `ConversationBufferWindowMemory`를 사용하는 것입니다.

````python
from langchain.chains.conversation.memory import ConversationBufferWindowMemory

# 대화 memory
conversational_memory = ConversationBufferWindowMemory(
    memory_key="chat_history", k=5, return_messages=True
)
````

아직 대화 memory를 agent에 연결하지 않았으므로 `conversational_memory` object는 비어 있습니다:

````python
conversational_memory.chat_memory.messages
````

우리와 agent 사이의 상호 작용을 memory에 수동으로 추가해야 합니다.

````python
conversational_memory.chat_memory.add_user_message(user_msg)
conversational_memory.chat_memory.add_ai_message(out["output"])

conversational_memory.chat_memory.messages
````

이제 우리의 `HumanMessage`와 agent의 `AIMessage` 응답, _두 개_의 message가 추가된 것을 볼 수 있습니다. 아쉽게도 이 message를 XML agent에 직접 보낼 수는 없습니다. 대신 다음 형식의 문자열을 전달해야 합니다:

```
Human: {human message}
AI: {AI message}
```

이를 처리할 간단한 `memory2str` 보조 함수를 작성하겠습니다:

````python
from langchain_core.messages.human import HumanMessage


def memory2str(memory: ConversationBufferWindowMemory):
    messages = memory.chat_memory.messages
    memory_list = [
        f"Human: {mem.content}" if isinstance(mem, HumanMessage) else f"AI: {mem.content}"
        for mem in messages
    ]
    memory_str = "\n".join(memory_list)
    return memory_str
````

````python
print(memory2str(conversational_memory))
````

이제 agent의 _state_ 부분을 처리하는 데 도움이 되는 `chat`이라는 다른 보조 함수를 작성하겠습니다.

````python
def chat(text: str):
    out = agent_executor.invoke({"input": text, "chat_history": memory2str(conversational_memory)})
    conversational_memory.chat_memory.add_user_message(text)
    conversational_memory.chat_memory.add_ai_message(out["output"])
    return out["output"]
````

이제 agent와 대화하기만 하면 이전 상호 작용의 context를 기억합니다.

````python
print(chat("was any red teaming done with the model?"))
````

핵심 정보가 빠진 후속 질문을 해도 대화 이력 덕분에 LLM이 context를 이해하고 이를 활용하여 검색 query를 조정합니다. 예를 들어 `red teaming`에 대해 질문하면서 `llama 2`는 언급하지 않았지만, Claude 3는 대화 이력을 기반으로 검색 query에 이 context를 추가하여 `"llama 2 red teaming"`으로 만들었습니다.
