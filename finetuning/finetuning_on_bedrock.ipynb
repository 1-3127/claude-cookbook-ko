## Dataset 준비
Bedrock finetuning용 dataset은 JSONL 파일(즉, 각 줄에 json object가 있는 파일)이어야 합니다.

JSONL 파일의 각 줄은 다음 구조의 JSON object여야 합니다:

```
{
  "system": "<optional_system_message>",
  "messages": [
    {"role": "user", "content": "user message"},
    {"role": "assistant", "content": "assistant response"},
    ...
  ]
}
```

- `system` field는 선택 사항입니다.
- Message가 최소 두 개 있어야 합니다.
- 첫 message는 "user"의 것이어야 합니다.
- 마지막 message는 "assistant"의 것이어야 합니다.
- User와 assistant message가 번갈아 나타나야 합니다.
- 불필요한 key는 허용되지 않습니다.


## 샘플 Dataset - JSON Mode
모든 질문에 JSON으로 응답하도록 모델을 학습시키는 샘플 dataset을 포함했습니다. 해당 dataset은 다음과 같습니다:

````python
import json

sample_dataset = []
dataset_path = "datasets/json_mode_dataset.jsonl"
with open(dataset_path) as f:
    for line in f:
        sample_dataset.append(json.loads(line))

print(json.dumps(sample_dataset[0], indent=2))
````

## Dataset을 S3에 업로드
Finetuning용 dataset은 s3에서 접근할 수 있어야 합니다. 이 데모에서는 샘플 dataset을 여러분이 관리하는 s3 bucket에 기록하겠습니다

````python
bucket_name = "YOUR_BUCKET_NAME"
s3_path = "json_mode_dataset.jsonl"

s3 = boto3.client("s3")
s3.upload_file(dataset_path, bucket_name, s3_path)
````

## Bedrock Finetuning Job 시작

Dataset이 준비되었으므로 `boto3`를 사용하여 finetuning job을 시작할 수 있습니다. 먼저 job에 필요한 몇 가지 parameter를 설정하겠습니다:

````python
# 설정
job_name = "anthropic-finetuning-cookbook-training"
custom_model_name = "anthropic_finetuning_cookbook"
role = "YOUR_AWS_SERVICE_ROLE_ARN"
output_path = f"s3://{bucket_name}/finetuning_example_results/"
base_model_id = (
    "arn:aws:bedrock:us-east-1::foundation-model/anthropic.claude-haiku-4-5-20251001-v1:0:200k"
)

# Hyperparameter
epoch_count = 5
batch_size = 4
learning_rate_multiplier = 1.0
````

그런 다음 `boto3`로 job을 시작할 수 있습니다

````python
bedrock = boto3.client(service_name="bedrock")
bedrock_runtime = boto3.client(service_name="bedrock-runtime")

bedrock.create_model_customization_job(
    customizationType="FINE_TUNING",
    jobName=job_name,
    customModelName=custom_model_name,
    roleArn=role,
    baseModelIdentifier=base_model_id,
    hyperParameters={
        "epochCount": f"{epoch_count}",
        "batchSize": f"{batch_size}",
        "learningRateMultiplier": f"{learning_rate_multiplier}",
    },
    trainingDataConfig={"s3Uri": f"s3://{bucket_name}/{s3_path}"},
    outputDataConfig={"s3Uri": output_path},
)
````

학습이 진행되는 동안 이를 사용하여 job 상태를 확인할 수 있습니다:

````python
# Job 상태 확인
status = bedrock.get_model_customization_job(jobIdentifier=job_name)["status"]
````

## Finetuning한 모델 사용하기!

Finetuning한 모델을 사용하려면 [Amazon Bedrock의 Provisioned Throughput으로 호스팅해야 합니다](https://docs.aws.amazon.com/bedrock/latest/userguide/model-customization-use.html). 모델의 Provisioned Throughput 준비가 완료되면 Bedrock API를 통해 모델을 호출할 수 있습니다.

````python
provisioned_throughput_arn = "YOUR_PROVISIONED_THROUGHPUT_ARN"
````

````python
bedrock = boto3.client("bedrock-runtime", region_name="us-east-1")
body = json.dumps(
    {
        "anthropic_version": "bedrock-2023-05-31",
        "max_tokens": 1000,
        "system": "JSON Mode: Enabled",
        "messages": [
            {
                "role": "user",
                "content": [{"type": "text", "text": "What is a large language model?"}],
            }
        ],
    }
)
response = bedrock_runtime.invoke_model(modelId=provisioned_throughput_arn, body=body)
body = json.loads(response["body"].read().decode("utf-8"))
````
