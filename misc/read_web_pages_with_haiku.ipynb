# Claude 3 Haiku로 Web Page 콘텐츠 요약
이 recipe에서는 URL이 주어졌을 때 web page의 내용을 fetch하고, Anthropic의 Claude API로 페이지 내용의 요약을 생성하는 방법을 배웁니다.

먼저 Anthropic library를 설치하겠습니다.

## 설정
먼저 필요한 library를 설치하고 API key로 Anthropic client를 설정합니다.

```python
# 필요한 library 설치
%pip install anthropic
```

```python
# 필요한 library import
from anthropic import Anthropic

# Claude API client 설정
client = Anthropic()
MODEL_NAME = "claude-haiku-4-5"
```

## 1단계: Web Page 콘텐츠 Fetch
먼저 제공된 URL을 사용해 web page의 내용을 fetch해야 합니다. 이를 위해 requests library를 사용합니다.

```python
import requests

url = "https://en.wikipedia.org/wiki/96th_Academy_Awards"
response = requests.get(url, timeout=30)

if response.status_code == 200:
    page_content = response.text
else:
    print(f"Failed to fetch the web page. Status code: {response.status_code}")
    exit(1)
```

## 2단계: Claude 입력 준비
다음으로 Claude API의 입력을 준비합니다. 페이지 내용과 이를 요약해 달라고 요청하는 프롬프트가 포함된 message를 만듭니다.

```python
prompt = (
    f"<content>{page_content}</content>Please produce a concise summary of the web page content."
)

messages = [{"role": "user", "content": prompt}]
```

## 3단계: 요약 생성
이제 Haiku를 호출해 web page 내용의 요약을 생성합니다.

```python
response = client.messages.create(model="claude-haiku-4-5", max_tokens=1024, messages=messages)

summary = response.content[0].text
print(summary)
```
